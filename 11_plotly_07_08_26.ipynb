Plotly Notes Plotly is an interactive plotting library for Python. It is commonly used to create dashboards, web-based visualizations, and charts with hover tools, zooming, and pan support.

What Plotly is used for Interactive charts Web dashboards Dynamic reporting Visual exploration of data Why it matters Plotly makes it easy to build charts that users can interact with directly. It is especially useful when the chart should be explored rather than just viewed as a static image.

1. Import Required Libraries

Plotly is an interactive visualization library that works well in notebooks and web apps.

Notes

plotly.express provides an easy way to create interactive charts. It works well with pandas DataFrames. Plotly charts can be zoomed, hovered, and explored without leaving the notebook.

```python
import pandas as pd
import plotly.express as px

print('Plotly is ready!')
```

2. Load and Inspect Sample Data

We will create a simple dataset and inspect it before creating charts.

Note

A DataFrame is a table of rows and columns. Before plotting, it is helpful to inspect the variables and their values. Plotly can visualize data directly from a pandas DataFrame.

```python
sales_df = pd.DataFrame({
    'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'sales': [120, 150, 170, 165, 210, 245],
    'product': ['A', 'A', 'A', 'B', 'B', 'B']
})

print(sales_df)
print('\nShape:', sales_df.shape)
print('\nHead:')
print(sales_df.head())
print('\nSummary:')
print(sales_df.describe())
```

3. Create a Line Plot

Line plots are ideal for showing change over time.

Notes px.line() creates an interactive line chart easily. Hovering over points shows exact values. This type of chart is helpful when you want to track trends over time.

```python
fig = px.line(
    sales_df,
    x='month',
    y='sales',
    title='Monthly Sales Trend',
    markers=True
)
fig.show()
```

4. Create a Bar Chart

Bar charts are useful for comparing values across categories.

Notes px.bar() is simple and interactive. This chart is best for comparing category totals. Interactive hover labels make it easier to inspect values quickly.

```python
fig = px.bar(
    sales_df,
    x='month',
    y='sales',
    title='Monthly Sales by Month',
    color='month'
)
fig.show()
```

5. Create a Scatter Plot

Scatter plots show the relationship between two numerical variables.

Notes

Scatter plots help show patterns, clusters, and outliers. If points move upward together, the relationship is positive. Plotly lets you explore each point with hover details.

```python
scatter_data = pd.DataFrame({
    'hours_studied': [1, 2, 3, 4, 5, 6, 7],
    'score': [40, 50, 55, 65, 75, 82, 90]
})

fig = px.scatter(
    scatter_data,
    x='hours_studied',
    y='score',
    title='Study Hours vs Score',
    size_max=15
)
fig.show()
```

6. Create a Histogram

A histogram shows the distribution of values in a single numeric variable.

Notes

Histograms show how values are spread across ranges. The number of bins affects the visual pattern. They are helpful for understanding skewness and concentration in the data.

```python
scores = [55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 82, 85, 90, 92]

fig = px.histogram(
    x=scores,
    nbins=6,
    title='Distribution of Scores'
)
fig.update_layout(xaxis_title='Score', yaxis_title='Frequency')
fig.show()
```

7. Create a Pie Chart

Pie charts show the share of each category in a total.

Notes Pie charts are useful when showing parts of a whole. They work best with a small number of categories. For detailed comparison, another chart type may be clearer.

```python
labels = ['Product A', 'Product B', 'Product C', 'Product D']
values = [25, 30, 20, 25]

fig = px.pie(
    names=labels,
    values=values,
    title='Product Sales Share'
)
fig.show()
```

8. Create an Interactive Dashboard-Style Chart

Plotly is especially strong for interactive and dashboard-like visuals.

Notes Plotly can display multiple series in one chart. Users can hover, zoom, and compare values quickly. This is ideal for interactive dashboards and live reporting.

```python
import numpy as np

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
product_a = [20, 25, 22, 30, 35, 40]
product_b = [15, 18, 20, 26, 28, 32]

fig = px.line(
    pd.DataFrame({
        'month': months,
        'Product A': product_a,
        'Product B': product_b
    }),
    x='month',
    y=['Product A', 'Product B'],
    title='Interactive Multi-Series Plot'
)
fig.update_layout(legend_title_text='Products')
fig.show()
```

9. Add Custom Styling and Layout

Plotly offers many styling options such as colors, titles, and axis formatting.

Notes

Styling improves readability and presentation quality. Plotly allows users to add information without cluttering the chart. Clean formatting helps the viewer focus on the story in the data.

```python
fig = px.line(
    sales_df,
    x='month',
    y='sales',
    title='Styled Plotly Chart',
    color_discrete_sequence=['#2E8B57']
)
fig.update_traces(mode='lines+markers', line=dict(width=3), marker=dict(size=9))
fig.update_layout(
    xaxis_title='Month',
    yaxis_title='Sales',
    template='plotly_white'
)
fig.show()
```

10. Save and Final Notes

You can export Plotly figures as HTML files for sharing.

Final Takeaway

Plotly is best when you want interactive and engaging data visualizations. It shines in dashboard and web-based work, where users can explore the data by hovering, zooming, and filtering. It is a strong choice when the goal is not just to show a chart, but to let the viewer explore it.

```python
fig = px.bar(
    sales_df,
    x='month',
    y='sales',
    title='Sales Chart for Export'
)
fig.write_html('plotly_sales.html')
print('Saved as plotly_sales.html')
fig.show()
```

Many Graph Examples Using the Seaborn Tips Dataset

Plotly is interactive, so it is ideal for exploring the tips dataset in a more dynamic way.

Notes Plotly is best when you want interactive graphs. Each example uses the tips dataset and focuses on a simple question. Interactive charts are helpful when the user wants to inspect details by hovering or zooming.

```python
fig = px.bar(
    sales_df,
    x='month',
    y='sales',
    title='Sales Chart for Export'
)
fig.write_html('plotly_sales.html')
print('Saved as plotly_sales.html')
fig.show()
import seaborn as sns
import plotly.express as px

# Load dataset
tips = sns.load_dataset('tips')
print(tips.head())

# 1. Line chart: average bill by day
avg_bill = tips.groupby('day', as_index=False)['total_bill'].mean()
fig = px.line(avg_bill, x='day', y='total_bill', title='1. Line Chart: Average Bill by Day', markers=True)
fig.show()

# 2. Bar chart: average tip by day
avg_tip = tips.groupby('day', as_index=False)['tip'].mean()
fig = px.bar(avg_tip, x='day', y='tip', title='2. Bar Chart: Average Tip by Day')
fig.show()

# 3. Scatter plot: total bill vs tip
fig = px.scatter(tips, x='total_bill', y='tip', color='sex', title='3. Scatter Plot: Total Bill vs Tip')
fig.show()

# 4. Histogram: total bill distribution
fig = px.histogram(tips, x='total_bill', nbins=20, title='4. Histogram: Total Bill Distribution')
fig.show()

# 5. Box plot: total bill by day
fig = px.box(tips, x='day', y='total_bill', title='5. Box Plot: Total Bill by Day')
fig.show()

# 6. Bar chart: average bill by sex
sex_bill = tips.groupby('sex', as_index=False)['total_bill'].mean()
fig = px.bar(sex_bill, x='sex', y='total_bill', title='6. Bar Chart: Average Bill by Sex')
fig.show()

# 7. Scatter plot: bill vs party size
fig = px.scatter(tips, x='total_bill', y='size', color='day', title='7. Scatter Plot: Bill vs Party Size')
fig.show()

# 8. Violin plot: bill by smoker status
fig = px.violin(tips, x='smoker', y='total_bill', box=True, title='8. Violin Plot: Total Bill by Smoker Status')
fig.show()

# 9. Pie chart: day share
fig = px.pie(tips, names='day', title='9. Pie Chart: Day Share')
fig.show()

# 10. Heatmap: correlation matrix
corr = tips[['total_bill', 'tip', 'size']].corr().round(2)
fig = px.imshow(corr, text_auto=True, color_continuous_scale='RdBu', title='10. Heatmap: Correlation Matrix')
fig.show()

# 11. Line chart: tip by day
fig = px.line(tips.groupby('day', as_index=False)['tip'].mean(), x='day', y='tip', title='11. Line Chart: Average Tip by Day', markers=True)
fig.show()

# 12. Bar chart: total bill by time
fig = px.bar(tips.groupby('time', as_index=False)['total_bill'].mean(), x='time', y='total_bill', title='12. Bar Chart: Average Bill by Meal Time')
fig.show()
```
